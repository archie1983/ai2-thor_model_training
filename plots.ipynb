```python
import matplotlib.pyplot as plt

def memory_needed(param_count = 7.0*10**9, quantization = 6.0):
    return round(((param_count * 4.0) / (32.0 / quantization)) * 1.2)

# LLM names and their memory requirements (in GB)
llms = ["24B-16bit", "7.04B-16bit", "8B-6bit", "7.3B-6bit", "7B-6bit", "7.3B-4bit", "1.87B-16bit"]
memory_requirements = [memory_needed(24.0, 16), 
                       memory_needed(7.04, 16), 
                       memory_needed(8.0, 6), 
                       memory_needed(7.3, 6), 
                       memory_needed(7.0, 6), 
                       memory_needed(7.3, 4), 
                       memory_needed(1.87, 16)]

# Create the bar chart
plt.figure(figsize=(8, 5))
plt.bar(llms, memory_requirements, color=['red', 'orange', 'green', 'green', 'green', 'green', 'green'])

# Add labels and title
plt.xlabel("Parameter Count - Quantization")
plt.ylabel("Memory Requirement (GB)")
#plt.title("Comparison of LLM Memory Requirements")
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show the values on top of bars
for i, value in enumerate(memory_requirements):
    plt.text(i, value + 1, f"{value} GB", ha='center', fontsize=10)

# Display the chart
plt.show()
```
